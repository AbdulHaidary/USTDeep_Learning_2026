```python
# ===============================
# Imports
# ===============================
import numpy as np
import matplotlib.pyplot as plt
import random

from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import (
    Dense, Dropout, Activation,
    Conv2D, MaxPooling2D, Flatten
)
from keras.utils import to_categorical


# ===============================
# Experiment setup
# ===============================
train_sizes = [1000, 5000, 10000, 20000, 40000, 60000]
fc_accuracies  = []
cnn_accuracies = []

print("Training sizes:", train_sizes)


# ===============================
# Load MNIST (fresh copy)
# ===============================
(X_train_raw, y_train_raw), (X_test_raw, y_test_raw) = mnist.load_data()
num_classes = 10


# ===============================
# Fully Connected preprocessing
# ===============================
X_train_fc = X_train_raw.reshape(-1, 784).astype("float32") / 255.0
X_test_fc  = X_test_raw.reshape(-1, 784).astype("float32") / 255.0

Y_train_fc = to_categorical(y_train_raw, num_classes)
Y_test_fc  = to_categorical(y_test_raw, num_classes)


# ===============================
# FC model builder
# ===============================
def make_fc_model():
    model = Sequential([
        Dense(512, input_shape=(784,)),
        Activation("relu"),
        Dropout(0.2),

        Dense(512),
        Activation("relu"),
        Dropout(0.2),

        Dense(num_classes),
        Activation("softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model


# ===============================
# Train FC models
# ===============================
for n in train_sizes:
    fc_model = make_fc_model()

    fc_model.fit(
        X_train_fc[:n],
        Y_train_fc[:n],
        epochs=5,
        batch_size=128,
        verbose=0
    )

    _, acc = fc_model.evaluate(X_test_fc, Y_test_fc, verbose=0)
    fc_accuracies.append(acc)

    print(f"FC  | n={n:>5} | acc={acc:.4f}")


# ===============================
# CNN preprocessing
# ===============================
X_train_cnn = X_train_raw.reshape(-1, 28, 28, 1).astype("float32") / 255.0
X_test_cnn  = X_test_raw.reshape(-1, 28, 28, 1).astype("float32") / 255.0

Y_train_cnn = to_categorical(y_train_raw, num_classes)
Y_test_cnn  = to_categorical(y_test_raw, num_classes)


# ===============================
# CNN model builder
# ===============================
def make_cnn_model():
    model = Sequential([
        Conv2D(32, (3,3), activation="relu", input_shape=(28,28,1)),
        MaxPooling2D((2,2)),

        Conv2D(64, (3,3), activation="relu"),
        MaxPooling2D((2,2)),

        Flatten(),
        Dense(128, activation="relu"),
        Dense(num_classes, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model


# ===============================
# Train CNN models
# ===============================
for n in train_sizes:
    cnn_model = make_cnn_model()

    cnn_model.fit(
        X_train_cnn[:n],
        Y_train_cnn[:n],
        epochs=5,
        batch_size=128,
        verbose=0
    )

    _, acc = cnn_model.evaluate(X_test_cnn, Y_test_cnn, verbose=0)
    cnn_accuracies.append(acc)

    print(f"CNN | n={n:>5} | acc={acc:.4f}")


# ===============================
# Plot results
# ===============================
plt.figure(figsize=(8,6))
plt.plot(train_sizes, fc_accuracies, marker="o", label="Fully Connected")
plt.plot(train_sizes, cnn_accuracies, marker="o", label="Convolutional")

plt.xlabel("Training Set Size")
plt.ylabel("Test Accuracy")
plt.title("MNIST Accuracy vs Training Size")
plt.legend()
plt.grid(True)
plt.show()
```
